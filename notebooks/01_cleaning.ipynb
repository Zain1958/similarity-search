```python
# Test different algorithms to see how they work on known mutashabihaat
from rapidfuzz.distance import Levenshtein
from rapidfuzz import fuzz

"""
#Verses 1:1 & 6:45
verse1 = df.loc[(df['chapter#'] == 6) & (df['verse#'] == 45), 'verse'].iloc[0] 
verse2 = df.iloc[1]["verse"]
"""

verse1 = df.loc[(df["chapter#"] == 2) & (df["verse#"] == 26), 'verse' ].iloc[0]
verse2 = df.loc[(df["chapter#"] == 74) & (df["verse#"] == 31), 'verse' ].iloc[0]

lscore = Levenshtein.distance(verse1, verse2)
pscore = fuzz.partial_ratio(verse1,verse2)

print(f"The levenshtein distance is: {lscore}")
print(f"The partial ratio distance is: {pscore}")
```

```python
## Experimenting with partial ratio alignment to find index of specific matching phrases
verse1 = df.loc[(df["chapter#"] == 2) & (df["verse#"] == 28), 'verse' ].iloc[0]
verse_list = [[22,66], [30, 40], [45,26]]
for x in verse_list:
    verse2 = df.loc[(df["chapter#"] == x[0]) & (df["verse#"] == x[1]), 'verse' ].iloc[0]
    result = fuzz.partial_ratio_alignment(verse1,verse2)
    print(f"The partial ratio distance between 2:28 and {x[0]}:{x[1]} is {result.score} ")
    print(f"Start index in verse: {result.dest_start}")
    print(f"End index in verse: {result.dest_end}")
```

```python
## Trying to reduce number of matches to only best matches with partial ratio
import heapq

verse1 = df.loc[(df["chapter#"] == 2) & (df["verse#"] == 28), 'verse' ].iloc[0]
score_array = []

for idx,x in df["verse"].items():
    score = fuzz.partial_ratio(verse1, x)
    if score > 50:
        score_array.append(score)

best_matches = heapq.nlargest(10, score_array)
print(best_matches)
for idx,x in df["verse"].items():
    score = fuzz.partial_ratio(verse1, x)
    xfound = False
    for y in best_matches:
        if score == y and xfound == False :
            print(x)
            xfound = True

```
